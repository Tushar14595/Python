```python
# Shuru mein bag khali hai
inventory = []

# Naya item jodne ka function
def add_item(item_name):
    inventory.append(item_name)
    print(item_name + " bag mein add ho gaya!")

# Bag ka saaman dekhne ka function
def show_inventory():
    print("--- Aapke Bag Mein Hai ---")
    for item in inventory:
        print(item)

# --- Button dabate hain (Functions ko call karte hain) ---
add_item("Sword")
add_item("Health Potion")
add_item("Magic Shield")

show_inventory()
```

```python
import random



rand = random.randint(1, 100)

chances = 5

while True:
  guess = int(input("Enter a number: "))

  if rand == guess:
        print("You Won!")
        break
  elif guess > rand:
        print("Too High")
        chances = chances - 1
  else:
        print("Too Low")
        chances = chances - 1

  if chances == 0:
        print("Game Over")
        print("The secret number was:", rand)
        break
```
